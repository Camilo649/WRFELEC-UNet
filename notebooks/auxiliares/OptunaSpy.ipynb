```python
import optuna
import pandas as pd

NOMBRE_MODELO = 'Ext U-Net Comb V2.1'

# Apuntamos a /tmp/ para ver los resultados en vivo mientras el .py corre
storage_monitor = f'sqlite:////tmp/optuna_entrenamiento_{NOMBRE_MODELO}.db'

study = optuna.load_study(
    study_name=f'optimizacion_entrenamiento_{NOMBRE_MODELO}',
    storage=storage_monitor
)

df = study.trials_dataframe()

print(f"Cantidad de pruebas completadas hasta ahora: {len(df)}")
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
display(df)
```

```python
# Filtramos solo los que terminaron bien (COMPLETE)
df_completados = df[df['state'] == 'COMPLETE'].copy()

# Ordenamos de menor a mayor (el mejor val_loss arriba)
df_top = df_completados.sort_values(by='value').reset_index(drop=True)

# Elegimos qué columnas mostrar para que quede prolijo
columnas_params = [col for col in df_top.columns if col.startswith('params_')]
columnas_mostrar = ['number', 'value'] + columnas_params

# Armamos el diccionario para renombrar columnas
nombres_limpios = {
    'number': 'Trial #', 
    'value': 'Mejor Val Loss'
}

# Le sacamos el 'params_' a hiperparámetros como lr, max_prob, etc.
for col in columnas_params:
    nombres_limpios[col] = col.replace('params_', '')

# Aplicamos los cambios al DataFrame final
df_top_limpio = df_top[columnas_mostrar].rename(columns=nombres_limpios)

print("\n🏆 TOP 10 MEJORES ENTRENAMIENTOS ORDENADOS POR VAL LOSS (ETAPA B):")
display(df_top_limpio.head(10))
```
